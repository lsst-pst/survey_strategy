## Major updates v4.3.1

Version 4.3.1 comes after minor updates from version 4.0, which represented the phase 3 recommendations.
The baseline is the nominal survey strategy, with 2x14.6s exposures per visit in grizy and a uniform rolling cadence with 3 cycles of rolling. The one-snap simulation is the nominal survey strategy as in the baseline, except 1x29.2s visits in grizy. 

Version 4.3.2 is almost the same as v4.3.1, however small updates in the underlying rubin_scheduler code prompted us to create this slight update to ensure new DDF simulations had a same-time baseline simulation comparison.

The changes from v4.0 include:
* the start date has been updated. This is the major content change - moving from May 1, 2025 to November 1, 2025. Specifically, this happened at v4.2.
* the DDF dither size was reduced to 0.2 degrees. This increases the cadence in the DDF fields and is consistent with requests from the science community. Specifically, at v4.1.
* template expiration time was updated from 300 days to 365 days. This is more in line with actual expectations of time between data releases, but does drive some change in g band coverage at the end of year one, where parts of the sky were inaccurately believing their template had expired and were rushing to get more coverage. Coverage over the sky within each year is now more even as a result. v4.2. 
* the DDF prescheduling code has been refactored to make it easier to read and thus for others to contribute DDF strategies. This did result in some minor changes in DDF coverage, but within noise expectations. v4.2
* some refactoring in various detailers and surveys may have a small impact on greedy (single visit during twilight) survey observations but should not significantly impact the overall survey. v4.2

Versions 4.1 and 4.2 were not intended to be widely released, and include various levels of updates on these internal code reconfigurations, but are included here for completeness. Statistically, these simulations should be very similar.
Version 4.3.1 fixes a problem with the dithering for the Euclid Deep Field South DDF, and replaces the v4.3 baseline.

```python
class TTotalNumberSSO(maf.BaseMoMetric):
    """Calculate the total number of objects of a given population
    expected at a given H value or larger.

    Operations on differential completeness values
    (or equivalent; fractions of the population is ok if
    still a differential metric result, not cumulative).

    Parameters
    ----------
    h_mark : `float`, optional
        The H value at which to calculate the expected total number of objects.
    dndh_func : function, optional
        The dN/dH distribution used calculate the expected population size.

    Returns
    -------
    nObj : `float`
        The predicted number of objects in the population.
    """

    def __init__(self, h_mark=22, dndh_func=maf.neo_dndh_granvik, dndh_args=None, **kwargs):
        self.h_mark = h_mark
        self.dndh_func = dndh_func
        self.dndh_args = dndh_args
        metric_name = "Nobj <= %.1f" % (h_mark)
        super().__init__(metric_name=metric_name, **kwargs)

    def run(self, metric_vals, h_vals):
        totals = maf.sum_over_h(metric_vals, h_vals, self.dndh_func, **self.dndh_args)
        n_obj = np.interp(self.h_mark, h_vals, totals)
        return n_obj
```
